```python
import torch

# Define three 2x3 matrices
A = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])

B = torch.tensor([[7, 8, 9],
                  [10, 11, 12]])

C = torch.tensor([[13, 14, 15],
                  [16, 17, 18]])

# Stack them horizontally (along dim=1, i.e., columns)
H = torch.cat([A, B, C], dim=1)

print(H)
H.shape  # torch.Size([2, 9])
```

```python
V = torch.cat([A, B, C], dim=0)
print(V)
V.shape
```

```python
T = torch.stack([A, B, C], dim=0)  # dim=0 means new axis at the front

print(T)
T.shape  # torch.Size([3, 2, 3])
```

```python
T[0]
```

### Random Tensors

Why random tensors?
They are important because many neural networks begin with random tensors, observe the data, and then adjust those numbers to better represent the data.

Torch random tensors: https://docs.pytorch.org/docs/stable/generated/torch.rand.html

```python
# Create a random tensor of size (3,4)

random_tensor = torch.rand(3, 4)
random_tensor
```

```python
random_tensor = torch.rand(1, 3, 4)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor = torch.rand(3, 4, 1)
random_tensor
```

```python
# create a random tensor with the size similar to an image tensor

random_image_size_tensor = torch.rand(size=(3, 224, 224)) # color channels (R, G, B), height, width
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

### Tensor datatypes

''Note'': Tensor datatypes is one of the 3 big errors you’ll run into with PyTorch & deep learning:

  1. Tensors not right datatype
  2. Tensors not right shape
  3. Tensors not on the right device

```python
# Float 32 tensor
float_32_tensor = torch.tensor(
    [3.0, 6.0, 9.0],
    dtype=None,       # what datatype is the tensor (e.g. float32 by default for floats)
    device=None,      # what device your tensor is on e.g. GPU, CPU, cuda
    requires_grad=False # track gradient?
)

float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tensor = torch.tensor(
    [3.0, 6.0, 9.0],
    dtype=torch.float16,       # what datatype is the tensor (e.g. float32 by default for floats)
    device=None,      # what device your tensor is on e.g. GPU, CPU, cuda
    requires_grad=False # track gradient?
)

print(f"Datatype of tensor: {float_16_tensor.dtype}")
print(f"Device tensor is on: {float_16_tensor.device}")
```

```python
product = float_16_tensor * float_32_tensor
# Normally, you might get into error if data types doesn't match
print(f"Datatype of tensor: {product.dtype}")
```
